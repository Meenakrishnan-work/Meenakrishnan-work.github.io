```python
"""Code cell <RODlzxoRJ_3b>
#%% [code]
#Personal mini chatbot"""
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

X = ["hello", "hi", "how are you",
     "what is your name", "who are you",
     "bye", "goodbye"]

y = ["greeting", "greeting", "greeting",
     "identity", "identity",
     "bye", "bye"]

v = TfidfVectorizer()
X = v.fit_transform(X)

model = LogisticRegression()
model.fit(X, y)

while True:
    msg = input("You: ")

    intent = model.predict(v.transform([msg]))[0]

    responses = {
        "greeting": "Hello! 😊",
        "identity": "I am your AI model!",
        "bye": "Goodbye! 👋"
    }

    print("AI:", responses[intent])

    if intent == "bye":
        break
```

```python
#Movie recommendation
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

movies = ["Avengers", "Spider Man", "Iron Man",
          "Titanic", "Notebook", "The Conjuring"]

genres = ["action", "action", "action",
          "romance", "romance", "horror"]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(genres)

model = LogisticRegression()
model.fit(X, genres)

user = input("What type of movie do you want? ")

genre = model.predict(vectorizer.transform([user]))[0]

print("\nAI recommends:")

for movie, movie_genre in zip(movies, genres):
    if movie_genre == genre:
        print("🎬", movie)
```
