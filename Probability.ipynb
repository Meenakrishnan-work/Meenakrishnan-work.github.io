Probability Distrubution

```python
import scipy.stats as stats
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

Assum T-shirt manufacturing defect, based on the historical data. Claims total defect is 20%

```python
#n is no. of sample
#pmf = probability mass function, x - no of sample to be tested, n - total no of t shirt, p - historical prob
n=50
p=0.20
stats.binom.pmf(3,n,p)
```

```python
x = np.arange(50)
```

```python
pval=stats.binom.pmf(x,n,p)
```

```python
plt.plot(x,pval,'o--')
plt.xticks(x)
plt.show()
#maximum value = n*p = 50*0.2 = 10
```

```python
#what is the probabilty of getting max 3 T-shirt defective
stats.binom.pmf(0,n,p)+stats.binom.pmf(1,n,p)+stats.binom.pmf(2,n,p)+stats.binom.pmf(3,n,p)
```

```python
stats.binom.cdf(3,n,p) #cumulative density function
```

```python
#what is the probabilty of atleast 3 T-shirt

1-stats.binom.cdf(3,n,p)
```

```python
#what is the probability of 5-10 (between)

n10 = 1-stats.binom.cdf(10,n,p)
n5 = stats.binom.cdf(5,n,p)
n = n10-n5
print (n)
```
